```python
s = input()
cnt = 0
for i in s:
    if i.isalpha():
        cnt +=1
    else:
        continue
print(cnt)
```

```python
import math
list = list(map(int,input().split()))
for n in list:
    s = 1
    for i in range(2, math.isqrt(n)+1):
        if (n%1==0):
            s+=i
            s+=n/1
    if (math.isqrt(n)**2==n):
        s-=1
    if s==n:
        print(n, end =' ')
```

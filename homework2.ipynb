Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.

```python
a,b,c=map(int,input().split())
if a+b>c and b+c>a and a+c>b:
    print("la do dai 3 canh tam giac")
    if a==b==c:
        print("Deu")
    elif a==c or a==b or b==c:
        print("can")
    elif a**2+b**2==c**2 or b**2+c**2==a**2 or a**2+c**2==b**2:
        print("vuong")
    else:
        print("thuong")
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax^2 + bx + c = 0)

```python
import math
a,b,c=map(float,input().split())
if a==0:
    if b==0:
        if c==0:
            print("ham vo so nghiem")
        else:
            print("ham vo nghiem")
    else:
        x=-b/c
        print(f"nghiem la{x}")
else:
    delta=b**2-4*a*c
    if delta<0:
        print("ham vo nghiem")
    elif delta>0:
        x=-b/2*a
        print(f"phuong trinh co nghiem kep la x={x} ")
    else:
        x1=(-b+math.sqrt(delta))/2*a
        x2=(-b-math.sqrt(delta))/2*a
        print(f"phuong trinh co 2 nghiem phan biet la x1={x1} x2={x2}")    
```
